#### The requirements of systematic trading: rule-based algorithmic trading

- `price`
- `noise reduced price`
- `(event) signal`
- <mark>position</mark>: `target`; `implemented` 
- `p&l`

````python
def detect_signal(df,
                  price_column = "Adj Close"):
    
    df_of_interest = df.copy()
    # noise-reduced price
    df_of_interest["sma_40"] =\
    (
        df_of_interest
        [price_column]
        .rolling(40,
                 min_periods = 1,
                 center = False)
        .mean()
    )

    df_of_interest["sma_100"] =\
    (
        df_of_interest
        [price_column]
        .rolling(100,
                 min_periods = 1,
                 center = False)
        .mean()
    )

    # (event) signal detection
    df_of_interest["Signal"] =\
    (
        df_of_interest["sma_40"] > df_of_interest["sma_100"]
    ).astype(int)

    return df_of_interest
````

#### The requirements of systematic trading: rule-based algorithmic trading

- `price`
- `noise reduced price`
- `(event) signal`
- <mark>position</mark>: `target`; `implemented` 
- `p&l`

````python
def compute_strategy_returns(df,
                             price_column = "Adj Close"):

    df = detect_signal(df, price_column = price_column)

    df["return"] = df[price_column].pct_change()

    df["position"] = df["Signal"].shift(1).fillna(0)

    df["strategy_return"] = df["return"] * df["position"]

    df["price_diff"] = df[price_column].diff()

    df["shares"] = 100 * df["position"]

    df["pnl"] = df["shares"] * df["price_diff"]

    return df
````

### Calculate and visualize the maximum drawdowns and the longest drawdown periods for `A`, `B`, and `C`.

### Below are the lines of code that lead to an answer:

````python
def compute_drawdown_metrics(df,
                             price_column = "Adj Close"):

    price_series = df[price_column].astype(float) # Step 1. make sure to take the price series

    wealth = price_series / price_series.iloc[0] # Step 2. normalization of the scale (our first value will be 1.0)
                                                 # Team, this is our buy-and-hold wealth index :)
    peak = wealth.cummax() # Step 3. running peak of our wealth curve

    drawdown =\
    ( # Step 4. the drawdown series --> drawdown will be negative numbers
        (wealth - peak) # numerator
        / 
        peak # denominator
    )

    in_drawdown = wealth < peak # Step 5. for Boolean masking!!
                                # Team, you will have True when we're underwater 
                                # that is, we are below the last peak

    # team, let us track the longest underwater streak!!
    longest = 0; current = 0; start_index = None; span = (None, None)
    
    # no drawdown yet # current ongoing dd
    for i, flag in enumerate(in_drawdown):
        if flag:
            current += 1
            if current == 1: start_index = i # this is when underwater streak starts
            if current > longest:
                longest = current; span = (start_index, i)
        else:
            current = 0 # reset when we are back to a new peak
                        # team, we are no longer underwater

    return wealth, drawdown, -drawdown.min() * 100, longest, span
                             # max drawdown in percentage format
````

###  <font color = blue> 👉 Question 5. </font> Report compound annual growth rate (CAGR) for `A`, `B`, and `C`.

### Below are the lines of code that lead to an answer:

````python
def compute_cagr(df,
                 price_column = "Adj Close"):
    # modularize all the elements
    start_value = df[price_column].iloc[0]
    
    end_value = df[price_column].iloc[-1]

    days =\
    (
        df["Date"].iloc[-1] # parse_dates --> datetime64
        - 
        df["Date"].iloc[0]
    ).days

    years = days/365.25

    return (end_value
            /
            start_value)**(1/years) - 1
````
