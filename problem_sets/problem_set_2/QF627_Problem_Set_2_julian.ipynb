* loan_amount < 5000 `low`

* loan_amount >= 15000 and loan_amount < 25000 `high`

```python
loan.loan_amnt.dtype
```

```python
def bin_loan_amt(loan_amount):
    if pd.isnull(loan_amount):
        return 'unknown'
    if loan_amount < 5_000:
        return 'low'
    elif loan_amount < 15_000:
        return 'medium'
    elif loan_amount < 25_000:
        return 'high'
    else:
        return 'very high'
```

> You should be able to write your own function which takes a categorical variable and plots the default rate segmented by purpose.

> Please **create your function** named `plot_segmented`

```python
import matplotlib.pyplot as plt

def plot_segmented(loan, feature):
    grouped = 1- loan.groupby(['purpose', feature])['loan_status'].mean().unstack()
    # Calculate the mean across columns for each 'purpose' and sort
    sorted_index = grouped.mean(axis=1).sort_values(ascending=False).index
    grouped_sorted = grouped.loc[sorted_index]

    plt.figure(figsize=(14, 7))
    grouped_sorted.plot(kind='bar', legend=True, ax=plt.gca())
    plt.title(f"Mean Loan Status by Purpose and {feature}")
    plt.ylabel("Mean Loan Status")
    plt.xlabel("Purpose")
    plt.tight_layout()
    plt.show()
```

#### Question 5.3 Write a function which takes in a categorical variable and computed the average default rate across the categories.

> You can also compute the `difference between the highest and the lowest default rate` across the categories, which is a decent metric indicating the effect of the varaible on default rate.

> Please create your own function named `diff_rate`.

```python
def diff_rate(df, feature):
    a = df[[feature, 'loan_status']].groupby(feature).mean()['loan_status']
    return float(a.max() - a.min()) * 100
```
