### Let's find: 
#### ***which occupations deliver the funniest TED talks on average?***

- for each talk, calculate the most frequent rating
- for each talk, clean the occupation data so that there's only one occupation per talk

##### You might want to count the number of funny ratings

````python
def all_ratings(row):
    max_dict = max(row['ratings_list'], key=lambda x: x['count'])
    return max_dict


df['most_frequent_rating'] = df.apply(all_ratings, axis=1)
df.head()
````

````python
import re

def clean_occupation(occupation):
    if pd.isna(occupation):
        return None
    s = str(occupation)
    parts = re.split(r'[^\w]+', s)
    tokens = [p.strip() for p in parts if p.strip()]
    if tokens:
        return tokens[0].lower()  # first token, lowercase
    return None

df['clean_occupation'] = df['speaker_occupation'].apply(clean_occupation)
df[['speaker_occupation', 'clean_occupation']].sample(20)
````

````python
df['clean_occupation'].value_counts().sample(20)
````

> "Funny" is not always the first dictionary in the list.

#### Please check ratings (not ratings_list) to see if "Funny" is always a rating type

````python
def funny_ratings(row):
    ratings = row['ratings']
    if 'Funny' in ratings:
        return True
    return False


df['funny_ratings'] = df.apply(funny_ratings, axis=1)
````

### Question 8. Let's write a custom function that gives us a `dictionary` that lists `talks` containing `funny` rating.

````python
def no_funny_ratings(df):
    ls = []
    for _, row in df.iterrows():
        rating = row['ratings']
        if 'Funny' not in rating:
            ls.append(row)
    return ls
````
