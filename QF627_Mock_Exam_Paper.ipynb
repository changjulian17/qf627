##  <font color = blue> 👉 Question 4. </font> Drawdown Analysis (20 points)

### For **Stock X (AAPL)** using the Bollinger Band strategy from Question 2:

### Tasks:

a) Calculate the cumulative returns (wealth index)

b) Calculate the drawdown series

c) Report the maximum drawdown (as a percentage)

d) Calculate the longest drawdown period (in days)

e) Visualize the wealth curve and drawdown series

f) Calculate the Calmar ratio (CAGR / Maximum Drawdown)

### Below are the lines of code that lead to an answer:

```python
def compute_drawdown_analysis(df, 
                              price_column = "Adj Close"):
    """
    Compute comprehensive drawdown metrics
    """
    price_series = df[price_column].astype(float)
    
    # Wealth index (normalized to 1.0)
    wealth = price_series / price_series.iloc[0]
    
    # Running peak
    peak = wealth.cummax()
    
    # Drawdown series
    drawdown = (wealth - peak) / peak
    
    # Track longest drawdown period
    in_drawdown = wealth < peak
    longest = 0
    current = 0
    start_idx = None
    span = (None, None)
    
    for i, flag in enumerate(in_drawdown):
        if flag:
            current += 1
            if current == 1:
                start_idx = i
            if current > longest:
                longest = current
                span = (start_idx, i)
        else:
            current = 0
    
    max_dd = -drawdown.min() * 100
    
    return wealth, drawdown, max_dd, longest, span
```

```python
# Apply drawdown analysis
```

```python
def compute_cagr(df, 
                 price_column = "Adj Close"):
    """
    Calculate Compound Annual Growth Rate
    """
    start_value = df[price_column].iloc[0]
    end_value = df[price_column].iloc[-1]
    
    days = (df.index[-1] - df.index[0]).days
    years = days / 365.25
    
    cagr = (end_value / start_value) ** (1 / years) - 1
    
    return cagr * 100
```

*Technical Indicators*:
- 21-day Simple Moving Average

- 14-day RSI (Relative Strength Index)
  - RSI = 100 - (100 / (1 + RS))
  - RS = Average Gain / Average Loss over 14 days

*Correlated Assets*:
- 5-day return of AAPL
- 5-day return of TSLA

### Tasks:

a) Engineer all the features listed above

b) Handle missing values (drop rows with NaN)

c) Split data: 70% training, 30% testing (use the first 70% for training chronologically)

d) Display the shape of training and testing sets

e) Show the first 5 rows of the feature matrix

### Below are the lines of code that lead to an answer:

```python
def calculate_rsi(prices, period = 14):
    """
    Calculate Relative Strength Index
    """
    delta = prices.diff()
    
    gain = delta.where(delta > 0, 0)
    loss = -delta.where(delta < 0, 0)
    
    avg_gain = gain.rolling(window = period, min_periods = 1).mean()
    avg_loss = loss.rolling(window = period, min_periods = 1).mean()
    
    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))
    
    return rsi
```

```python
# Biplot for PC1 and PC2
def create_biplot(pca_scores, pca_loadings, labels, pc1=0, pc2=1):
    """
    Create a biplot for PCA
    """
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Plot scores
    ax.scatter(pca_scores[:, pc1], pca_scores[:, pc2], alpha=0.3, s=10)
    
    # Plot loadings
    for i, label in enumerate(labels):
        ax.arrow(0, 0, 
                pca_loadings[i, pc1] * 3, 
                pca_loadings[i, pc2] * 3,
                head_width=0.1, head_length=0.1, 
                fc='red', ec='red', alpha=0.7)
        ax.text(pca_loadings[i, pc1] * 3.2, 
               pca_loadings[i, pc2] * 3.2,
               label, fontsize=12, ha='center', va='center')
    
    ax.set_xlabel(f'PC{pc1+1}')
    ax.set_ylabel(f'PC{pc2+1}')
    ax.set_title('PCA Biplot')
    ax.grid(True, alpha=0.3)
    
    return fig, ax
```
