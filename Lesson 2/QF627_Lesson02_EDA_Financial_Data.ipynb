````python
def get_multiple_stocks(tickers, start, end):

    def get_single_stock(ticker):
    
        return yf.download(ticker,
                           start, 
                           end)
    
    interim_DF = map(get_single_stock, tickers)
    
    return pd.concat(interim_DF,
                     keys = tickers,
                     names = ["ticker", "date"]
                    )["Close"]
````

### Moving average: 

> One of the analyses you can first look at is moving averages (MA), which is commonly used to identify trading opportunities. 

> It is calculated by taking the mean or average of the past data points of the prices. 

> It is called a moving average, as opposed to just an average, because the data set is continuously `moving`: the oldest data points are dropped from the data set to account for the new data as they become available.

> The moving average graph is a smooth line that follows the day-to-day values of the prices we are tracking but it has some lags. 

```python
(
    DF
    ["Series"]
    .rolling(window = `timeframe`)
    .mean()
)
```

````python
(
    stock_CLEANED
    [["NFLX"]]
    .head(6)
    # .rolling(window = 5) # 5 trading days ==> one week
    # .mean() # rolling averages
)
````

````python
(
    stock_CLEANED
    [["NFLX"]]
    .rolling(window = 5) # 5 trading days ==> one week
    .mean() # rolling averages
)
````

- Depending on the type of investor or trader (e.g., short-term vs. long-term; high risk vs. low risk trading), you may adjust your `moving time average` (10 days, 20 days, 50 days, 200 days, 1 year, 5 years, etc). 


- The two widely MAs that traders and investors used are 50-day MA and 200-day MA.

> Which moving average (50-day MA or 200-day MA) do you think will have a greater degree of lag?

#### ANSWER: 200-day MA

````python
stock_CLEANED["NFLX"].rolling(5).mean().plot(figsize = [16, 7]) # A Week

stock_CLEANED["NFLX"].rolling(22).mean().plot(figsize = [16, 7]) # A Month

stock_CLEANED["NFLX"].rolling(48).mean().plot(figsize = [16, 7]) # Three Months

stock_CLEANED["NFLX"].rolling(252).mean().plot(figsize = [16, 7]) # A Year
````

## Least squares regression of returns (beta)

$$
    GS = b_0 + \widehat{b_1}JPM + \epsilon 
$$

> Let's regress the returns of GS onto JPM, using `Ordinary least squares` method.

````python
regress_GS_onto_JPM =\
(
    smf
    .ols(formula = "GS ~ JPM", # Ordinary Least Squares
         data = dpc_method_pct_change)
    .fit() # fit() calculates
)
````

````python
print(regress_GS_onto_JPM
      .summary() # returns three sections of information
     )
````

````python
regress_GS_onto_UAL =\
(
    smf
    .ols(formula = "GS ~ UAL", # Ordinary Least Squares
         data = dpc_method_pct_change)
    .fit() # fit() calculates
)
````

````python
print(regress_GS_onto_UAL
      .summary() # returns three sections of information
     )
````

````python
regress_GS_onto_JPM.params[1]
````

````python
regress_GS_onto_UAL.params[1]
````

````python
# !pip install stargazer
````

````python
from stargazer.stargazer import Stargazer
        # NOTE that Capitalized S
````

````python
regression_table_to_report =\
    Stargazer([regress_GS_onto_JPM,
               regress_GS_onto_UAL] # plug-in list of models
             )
````

````python
regression_table_to_report\
    .title("Regression Models between GS Daily Returns & JPM, UAL Daily Returns")

regression_table_to_report\
    .custom_columns(["Model GS-JPM", "Model GS-UAL"],
                    [1, 1]
                   )

regression_table_to_report\
    .show_model_numbers(True)
````
